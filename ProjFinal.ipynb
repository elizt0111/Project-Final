#### Question 1: What is the distribution by county of waterfalls that are sourced from the Delaware River?

#### Question 2: Out of the municipalities with more than 3 waterfalls, which ones have the biggest on average?

I first found out which towns have over 3 waterfalls.

```python
town_counts = wf['town'].value_counts()
over3values = town_counts[town_counts > 3]
over3values
```

I then added all of these town names to a list.

```python
town_list = ['Hardwick Twp.', 'Walpack Twp.', 'Chester Twp.', 'Washington Twp.', 'West Milford Twp.', 'Alpine Borough', 'Kingwood Twp.', 'Mahwah Twp.', 'Bloomingdale Borough', 'Wantage Twp.', 'Little Falls']
town_list
```

I used this list to create a dataset that only contained the towns in the list.

```python
wf_over3 = wf[wf['town'].isin(town_list)]
wf_over3
```

I then made two variables that calculate the mean of all waterfalls in each of the selected towns. One shows the mean of the total waterfall heights, while the other shows the mean of the largest cascade.

```python
over3total = wf_over3.groupby('town')['height_total'].mean().sort_values()
over3casc = wf_over3.groupby('town')['height_cascade'].mean().sort_values()
over3 = wf_over3.groupby('town')['height_total', 'height_cascade'].mean()
```

I was having trouble getting my plot to work but learned that the index needs to be reset before plotting.

```python
over3_plot = over3_reset = over3.reset_index()

sns.barplot(data = over3_plot, x = 'town', y = 'height_total')
plt.xticks(rotation = 50, ha = 'right')
plt.title('Average Height of Waterfalls by Town')
plt.show()
```

```python
sns.barplot(data = over3_plot, x = 'town', y = 'height_cascade')
plt.xticks(rotation = 50, ha = 'right')
plt.title('Average Height of Waterfall Cascades by Town')
plt.show()
```

On average, Alpine has the largest waterfalls, while Kingwood has the largest cascades.

#### Question 4: Is there a correlation between waterfall type and total height?

I first wanted to determine which types of waterfalls had enough data to work with. I only want to analyze types that have more than 5 records. I decided to only use the types listed under type1, since this is the primary classification for the waterfall.

```python
wf.value_counts('type1')
```

Only 3 types of waterfalls have over 5 records, and 2 of them are barely making the cut. Cascade waterfalls have overwhelmingly more data. Let's see how this plays out.

I now have to filter the dataset so it only shows waterfall types with over 5 records.

```python
q5 = wf['type1'].value_counts()
over5 = q5[q5 > 5]
top3types = ['Cascade', 'Multi-step waterfall', 'Chute']
```

```python
top3 = wf[wf['type1'].isin(top3types)]
```

I then made a boxplot graph, hoping this will demonstrate correlation between the two variables:

```python
sns.boxplot(data = top3, x = 'height_total', y = 'type1')
plt.title('Distribution of Total Height (ft) Over Waterfall Types')
plt.show()
```

There may be a slight correlation between height and waterfall type. There is not much of a dramatic difference between heights, but with what were working with, it seems like the two variables are slightly correlated.
